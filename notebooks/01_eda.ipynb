```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

%matplotlib inline

df = pd.read_csv(
    "../data/MachineLearningRating_v3.txt",
    sep="|"
)

df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

Missing values were checked across all columns. Numerical missing values may be handled using median imputation, while categorical missing values can be handled using mode replacement or marked as unknown depending on business context.


```python
df.describe()
```

```python
df["LossRatio"] = (
    df["TotalClaims"] /
    df["TotalPremium"]
)

overall_loss_ratio = (
    df["TotalClaims"].sum() /
    df["TotalPremium"].sum()
)

print("Overall Loss Ratio:", overall_loss_ratio)
```

```python
plt.figure(figsize=(8,5))

df["TotalPremium"].hist(bins=30)

plt.title("Distribution of Total Premium")
plt.xlabel("Total Premium")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

df["TotalClaims"].hist(bins=30)

plt.title("Distribution of Total Claims")
plt.xlabel("Total Claims")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(x=df["TotalClaims"])

plt.title("Outliers in Total Claims")

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12,8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    x="TotalPremium",
    y="TotalClaims",
    data=df
)

plt.title("Total Premium vs Total Claims")

plt.show()
```

```python
province_analysis = df.groupby("Province")["TotalPremium"].mean()

province_analysis.sort_values().plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Average Premium by Province")

plt.ylabel("Average Premium")

plt.show()
```

```python
vehicle = df.groupby("VehicleType")[["TotalClaims","TotalPremium"]].sum()

vehicle["LossRatio"] = (
    vehicle["TotalClaims"] / vehicle["TotalPremium"]
)

vehicle["LossRatio"].sort_values().plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Loss Ratio by Vehicle Type")

plt.ylabel("Loss Ratio")

plt.show()
```

```python
gender = df.groupby("Gender")[["TotalClaims","TotalPremium"]].sum()

gender["LossRatio"] = (
    gender["TotalClaims"] / gender["TotalPremium"]
)

gender["LossRatio"].plot(kind="bar")

plt.title("Loss Ratio by Gender")

plt.show()
```

```python
df["TransactionMonth"] = pd.to_datetime(df["TransactionMonth"])

monthly_claims = df.groupby(
    df["TransactionMonth"].dt.to_period("M")
)["TotalClaims"].sum()

monthly_claims.plot(
    figsize=(12,6)
)

plt.title("Monthly Claims Trend")

plt.xticks(rotation=45)

plt.show()
```

```python
vehicle_claims = df.groupby("Make")["TotalClaims"].mean()

print("Lowest Claim Vehicles")
print(vehicle_claims.sort_values().head())

print("Highest Claim Vehicles")
print(vehicle_claims.sort_values().tail())
```

```python
print(df.columns.tolist())
```

```python
vehicle_claims = df.groupby("make")["TotalClaims"].mean()

print("Lowest Claim Vehicles")
print(vehicle_claims.sort_values().head())

print("\nHighest Claim Vehicles")
print(vehicle_claims.sort_values().tail())
```

```python
province_claims = df.groupby("Province")["TotalClaims"].mean().sort_values(ascending=False)

province_claims.head(10).plot(
    kind="bar",
    figsize=(12,6)
)

plt.title("Top Provinces by Average Claims")
plt.ylabel("Average Claims")

plt.show()
```

# Final Insights

## Key Findings

* The portfolio loss ratio provides insight into overall insurance profitability.
* Certain provinces demonstrate higher insurance risk than others.
* Vehicle type and vehicle make significantly influence claim behavior.
* Outliers exist in TotalClaims and CustomValueEstimate.
* Temporal analysis shows variations in claims over time.
* Gender and geographic segmentation may support risk-based pricing.

## Recommendations

* Introduce differentiated pricing for low-risk customer groups.
* Monitor high-claim outlier customers carefully.
* Use geographic and vehicle-based segmentation in predictive models.
* Build claim probability and severity models in future tasks.
